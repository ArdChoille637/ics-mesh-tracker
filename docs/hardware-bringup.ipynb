# ICS Mesh Tracker - System Progress

This notebook documents the progress made in deploying and testing the ICS (Incident Command System) Mesh Tracker.

## 1. Hardware Fleet Deployment
We successfully flashed a 4-node ESP-NOW mesh network:
- **1x Gateway Node** (`XIAO ESP32S3`)
- **1x Team Lead Node** (`XIAO ESP32S3`)
- **2x Field Nodes** (`1x XIAO ESP32S3`, `1x Arduino Nano ESP32`)


## 2. Firmware Fixes (Core 3.0 Migration)
The original firmware was built for an older ESP32 Arduino core. We upgraded it to support the modern `pioarduino` core (ESP-IDF v5) by patching:
1. **PlatformIO Configuration**: Switched platform to `https://github.com/pioarduino/platform-espressif32/releases/download/stable/platform-espressif32.zip`.
2. **Type Renaming**: Fixed the `esp_now_recv_info_t` struct renaming that changed in Core 3.0.
3. **Serial Interfaces**: Updated `GatewayBridge` to accept a generic `Stream&` instead of hardcoding `HardwareSerial`, fixing compatibility with the ESP32-S3's native `HWCDC` USB serial port.
4. **Added Arduino Nano ESP32**: Added a custom `[env:field_nano_esp32]` config with `upload_protocol = esptool` to bypass Mac DFU issues.

## 3. SBC Gateway Server & Dashboard
The SBC (Single Board Computer) gateway was deployed using Python and FastAPI.

- **Environment Setup**: Created a virtual environment and installed `fastapi`, `pyserial`, `websockets`, `numpy`, and `scipy`.
- **Execution**: Ran the server bound to the Gateway node on `/dev/cu.usbmodem2101`.
- **Command Map**: Successfully served the live command map on `http://localhost:8000`.

## 4. Live Test Results
During live testing of the mesh, the following was observed:

- **Node Registration**: All powered nodes successfully established connections via ESP-NOW and were relayed to the gateway.
- **Telemetry**: Nodes reported `1s` fresh ping times.
- **Status Updates**: PAR (Personnel Accountability Report) status updates flowed correctly (e.g., `OK`, `EMERGENCY`, `MAYDAY`).
- **ICS-214 Logs**: Custom text messages (like `Test` and `Testing`) sent from the captive portal correctly surfaced on the dashboard log.
